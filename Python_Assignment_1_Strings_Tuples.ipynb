## 1. String Concatenation

```python
string1 = "Hello "

name = input("Enter your Name: ")

string2 = string1 + name

print(string2)

string3 = ", welcome to Python programming"

string2 = string2 + string3

print(string2)
```

## 2. String Slicing and Indexing

```python
text = string2

```

```python
 text[0]
```

```python
text[-1]
```

```python
text[:5]
```

```python
text[-11:]
```

```python
text[::-1]
```

```python
text[text.index("Python"):text.index("Python")+6]
```
